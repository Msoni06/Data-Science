```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Unemployment_Rate_upto_11_2020.csv')
df.columns = ['Region', 'Date', 'Frequency', 'Estimated Unemployment Rate', 
              'Estimated Employed', 'Estimated Labour Participation Rate', 
              'Region.1', 'Longitude', 'Latitude']
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
print(df.head())
```

```python
plt.figure(figsize=(14, 7))
sns.lineplot(data=df, x='Date', y='Estimated Unemployment Rate')
plt.title('India\'s Unemployment Rate Trend', fontsize=16)
plt.xlabel('Date')
plt.ylabel('Estimated Unemployment Rate (%)')
plt.axvline(pd.to_datetime('2020-03-31'), color='red', linestyle='--', label='COVID-19 Lockdown Start')
plt.legend()
plt.show()
```

```python
state_unemployment = df.groupby('Region')['Estimated Unemployment Rate'].mean().sort_values(ascending=False)

plt.figure(figsize=(12, 10))

sns.barplot(y=state_unemployment.index, x=state_unemployment.values)

plt.title('Average Unemployment Rate by State', fontsize=16)
plt.xlabel('Estimated Unemployment Rate (%)')
plt.ylabel('State')

plt.show()
```
